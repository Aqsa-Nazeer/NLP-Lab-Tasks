```python
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        print(token, "→ Alphabetic (WORD)")
    elif token.isdigit():
        print(token, "→ Numeric (NUMBER)")
    else:
        print(token, "→ Mixed")
```

```python
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for word in words:
    if word.isalpha():
        pure_words.append(word)
    elif word.isdigit():
        pure_numbers.append(word)
    else:
        mixed_tokens.append(word)

print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)
```

```python
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"


tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for token in tokens:
    label = validate_token(token)
    print(token, "→", label)
```

## Task 08: Functions and Lambda for NLP Pipelines

```python
def clean(text):
    text = text.strip()
    text = text.lower()
    return text


clean_lambda = lambda text: text.strip().lower()


text = "  HELLO Natural Language Processing  "

print("Regular function:", clean(text))
print("Lambda function:", clean_lambda(text))
```

```python
def tokenize(text):
    return text.split()


def remove_short(tokens, n):
    return [token for token in tokens if len(token) > n]


sentence = "I love working with NLP tools in Python daily"

tokens = tokenize(sentence)
result = remove_short(tokens, 3)

print("Tokens:", tokens)
print("After removing short words:", result)
```

```python
def word_lengths(tokens):
    return {word: len(word) for word in tokens}


tokens = ["language", "model", "ai", "processing", "deep"]

result = word_lengths(tokens)

print("Word lengths:", result)
```

```python
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered_tokens = remove_short(tokens, min_len)
    return word_lengths(filtered_tokens)


sentence1 = "  Natural Language Processing is very interesting  "
sentence2 = "  Deep Learning helps machines understand text  "

result1 = nlp_pipeline(sentence1, 3)
result2 = nlp_pipeline(sentence2, 3)

print("Pipeline Output 1:", result1)
print("Pipeline Output 2:", result2)
```

## Task 09: map(), filter(), sorted() and Counter

```python
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lowercase_words = list(map(lambda word: word.lower(), words))
print("Lowercase words:", lowercase_words)

word_lengths = list(map(lambda word: len(word), words))
print("Word lengths:", word_lengths)
```

```python
words_long = list(filter(lambda word: len(word) > 3, words))
print("Words longer than 3 characters:", words_long)

words_with_t = list(
    filter(lambda word: "t" in word.lower(), words)
)
print("Words containing 't':", words_with_t)
```

```python
alphabetical = sorted(words)
print("Alphabetical:", alphabetical)

length_ascending = sorted(words, key=len)
print("Length ascending:", length_ascending)

length_descending = sorted(words, key=len, reverse=True)
print("Length descending:", length_descending)
```

```python
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"

word_list = sentence.split()

freq = Counter(word_list)

print("Counter:", freq)

print("3 most common words:", freq.most_common(3))

print("Frequency of python:", freq["python"])

print("Total number of words:", len(word_list))

print("Unique vocabulary size:", len(freq))
```
